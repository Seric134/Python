```python
# Question 2: String slicing and indexing

text = final_string

# a. First character
print(text[0])

# b. Last character
print(text[-1])

# c. First five characters
print(text[:5])

# d. Last eleven characters
print(text[-11:])

# e. Reverse the string
print(text[::-1])

# f. Extract the word Python
start = text.index("Python")
print(text[start:start+6])
```

```python
# Question 4: Tuple Operations

# Create two tuples
t1 = (10, 20, 30)
t2 = (40, 50, 60)

# a. Concatenate both tuples
t_combine = t1 + t2
print("a. Concatenated tuple:", t_combine)

# b. Repeat the elements three times
print("b. Repeated tuple:", t_combine * 3)

# c. Access the third element
print("c. Third element:", t_combine[2])

# d. Access the first three elements
print("d. First three elements:", t_combine[:3])

# e. Access the last three elements
print("e. Last three elements:", t_combine[-3:])
```
