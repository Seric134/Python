```python
age_list = [24, 25, 26, 27, 28]

print(age_list)

# Explanation:
# A list is used to store multiple elements in a single variable.
# Square brackets [] are used to create a list.
# Here, age_list contains five integer values.
```

```python
name_list = ["Arun", "Rahul", "Priya", "Kiran", "Anu"]

print(name_list)

# Explanation:
# String elements are enclosed within quotation marks.
# Each element is separated by a comma.
```

```python
name_list.append("Yazhini")

print(name_list)

# Explanation:
# append() adds a new element at the end of the list.
```

```python
age_list.insert(2, 30)

print(age_list)

# Explanation:
# insert(index, value) adds an element at a specific position.
# Index 2 means the third position because Python indexing starts from 0.
```

```python
name_list.remove("Yazhini")

print(name_list)

# Explanation:
# remove() deletes the specified element from the list.
# It removes the first matching occurrence.
```

```python
age_list.pop()

print(age_list)

# Explanation:
# pop() removes the last element by default.
# It can also remove an element using its index.
```

```python
age_list.extend([29, 30, 26])

print(age_list)

# Explanation:
# extend() adds multiple elements from another iterable.
# Unlike append(), it adds individual elements rather than
# adding the entire second list as one nested element.
```

```python
age_list.sort(reverse=True)

print(age_list)

# Explanation:
# sort() arranges elements in ascending order by default.
# reverse=True changes the sorting order to descending.
```

```python
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))

# Explanation:
# max() returns the largest value.
# min() returns the smallest value.
# sum() calculates the total of all numerical elements.
```

```python
print(name_list[0])

# Explanation:
# Index 0 represents the first element in a Python list.
```

```python
print(name_list[-1])

# Explanation:
# Negative indexing starts counting from the end.
# Index -1 represents the last element.
```

```python
print(name_list[2:5])

# Explanation:
# Slicing extracts a portion of a list.
# The starting index is included, but the ending index is excluded.
# Therefore, [2:5] returns elements at indexes 2, 3 and 4.
```

```python
print(name_list[::-1])

# Explanation:
# The slicing syntax [::-1] reverses the order of the list.
# It does not modify the original list.
```

```python
student_marks = {
    "Arun": 85,
    "Rahul": 76,
    "Priya": 92,
    "Kiran": 68,
    "Anu": 89
}

print(student_marks)

# Explanation:
# A dictionary stores information in key-value pairs.
# Here, student names are keys and their marks are values.
# Curly brackets {} are used to create a dictionary.
```

```python
print("Priya's mark:", student_marks["Priya"])

# Explanation:
# We can access a dictionary value using its key.
# Here, "Priya" is the key, and 92 is its corresponding value.
```

```python
student_marks["Janani"] = 80

print(student_marks)

# Explanation:
# Assigning a value to a new key adds a new key-value pair.
# Here, Janani is added with a mark of 80.
```

```python
student_marks["Rahul"] = 82

print(student_marks)

# Explanation:
# Assigning a new value to an existing key updates its value.
# Rahul's previous mark of 76 is now changed to 82.
```

```python
print("Student Names:", student_marks.keys())

print("Student Marks:", student_marks.values())

print("Student Details:", student_marks.items())

# Explanation:
# keys() returns all dictionary keys (student names).
# values() returns all dictionary values (student marks).
# items() returns all key-value pairs together.
```

```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print(my_set)

# Explanation:
# A set is a collection of unique elements.
# Duplicate elements are automatically removed.
# Therefore, even though 'a' appears three times and 'i'
# appears twice, each appears only once in the set.
# Sets do not guarantee a particular display order.
```

```python
my_set[4] = 's'

# Explanation:
# This statement produces a TypeError.
# Sets are unordered collections and do not support indexing.
# Therefore, we cannot access or modify an element using [4].
# Sets are mutable, but individual elements cannot be changed by index.
```

```python
my_set.add('s')

print(my_set)

# Explanation:
# add() inserts a new element into a set.
```

```python
set1 = {1, 3, 5, 7, 9}

set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)

# Explanation:
# Two sets are created with different integer elements.
# Notice that both sets contain 3 and 5.
```

```python
print("Union:", set1.union(set2))

print("Intersection:", set1.intersection(set2))

# Explanation:
# UNION combines all unique elements from both sets.
# INTERSECTION returns only the common elements.
# Here, the common elements are 3 and 5.
```

```python
print("Union:", set1.union(set2))

print("Intersection:", set1.intersection(set2))

# Explanation:
# UNION combines all unique elements from both sets.
# INTERSECTION returns only the common elements.
# Here, the common elements are 3 and 5.
```

```python
score = float(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid Score! Please enter a score between 0 and 10.")

elif score > 7:
    print("Performance Category: Above Average")
    print("Excellent work! Keep maintaining your performance.")

elif score >= 4:
    print("Performance Category: Average")
    print("Good effort! Keep practising to improve your performance.")

else:
    print("Performance Category: Below Average")
    print("Need to improve your performance. Consistent practice will lead to better results.")


# Explanation:
# input() accepts the score entered by the user.
# float() converts the input into a numerical value.
#
# IF checks whether the score is outside the valid range.
# OR checks whether either of the two conditions is true.
#
# ELIF means else if. It checks another condition
# when the previous condition is false.
#
# Score greater than 7 = Above Average.
# Score between 4 and 7 (inclusive) = Average.
# Score below 4 = Below Average.
#
# ELSE executes when none of the previous conditions are true.
```
